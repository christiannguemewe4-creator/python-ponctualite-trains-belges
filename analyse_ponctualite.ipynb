```python
ponctualite_ligne = df_arr.groupby("LINE_NO_ARR")["a_l_heure"].agg(["mean", "count"])
ponctualite_ligne

```

```python
lignes_fiables.sort_values("mean").head(10)
```

```python
lignes_fiables["nb_retards"] = (1 - lignes_fiables["mean"]) * lignes_fiables["count"].round()
lignes_fiables.sort_values("nb_retards", ascending=False).head(10)
```

```python
total_retards = ((1 - ponctualite_ligne["mean"]) * ponctualite_ligne["count"]).sum()
lignes_fiables["part_retards"] = (lignes_fiables["nb_retards"] / total_retards * 100).round(1)
lignes_fiables.sort_values("nb_retards", ascending=False).head(10)
```

## Conclusion : gares et lignes

Les lignes ont été filtrées pour garder celles avec au moins 1 000 arrivées par mois, afin que leur chiffre soit fiable statistiquement.

Seulement 5 lignes sur 148 (25, 36N, 36, 96 et 37) concentrent à elles seules 29 % des arrivées en retard du réseau belge. Les lignes 25, 36 et 96 sont de grands axes autour de Bruxelles, très fréquentés.

La ligne 37 (Liège–Verviers–Welkenraedt, vers l'Allemagne) est la priorité : elle cumule un mauvais taux (16,2 % de retard) et un gros volume (43 352 arrivées, soit environ 7 000 arrivées en retard sur le mois). Elle dessert aussi la plupart des gares les moins ponctuelles du pays : Hergenrath, Chênée, Welkenraedt, Verviers-Palais. Ses trains internationaux pourraient arriver déjà en retard d'Allemagne, à vérifier.

Recommandation : concentrer les efforts d'abord sur la ligne 37, puis sur les quatre grands axes (25, 36N, 36 et 96).

# Question 4: Les retards s’aggravent-ils le long du trajet ?

1- observation d'un seul trajet

```python
premier = df_arr.iloc[0]
un_train = df_arr[(df_arr["TRAIN_NO"] == premier["TRAIN_NO"]) & (df_arr["DATDEP"] == premier["DATDEP"])]
un_train[["PTCAR_LG_NM_NL", "PLANNED_TIME_ARR", "DELAY_ARR"]]
```

2- Numérotation des arrêts

```python
df_arr["num_arret"] = df_arr.groupby(["TRAIN_NO", "DATDEP"]).cumcount() + 1
df_arr["num_arret"]
```

3-le retard moyen par numéro d’arrê

```python
retard_moyen = df_arr.groupby("num_arret")["DELAY_ARR"].agg(["mean", "count"])
retard_moyen

```

```python
arret_fiables = retard_moyen[retard_moyen["count"] >= 1000].copy()
arret_fiables
```

```python
(arret_fiables["mean"] / 60).plot(kind="line")
plt.title("Retard moyen selon le numéro d'arrêt – septembre 2026")
plt.ylabel("Retard moyen (minutes)")
plt.xlabel("Numéro d'arrêt dans le trajet")
plt.show()
```

## Conclusion : évolution du retard le long du trajet

Oui, les retards s'aggravent le long du trajet, mais modérément : environ 45 secondes de plus entre le 1er et le 18e arrêt.

Entre le 1er et le 18e arrêt, le retard moyen passe de 1,6 à environ 2,35 minutes, probablement à cause de la montée des voyageurs, des croisements et des attentes de correspondances. Après le 18e arrêt, la courbe se stabilise et zigzague entre 2,15 et 2,35 minutes jusqu'au 39e arrêt, puis monte jusqu'à un pic de 2,66 minutes au 52e arrêt.

Après le 52e arrêt, le retard chute jusqu'à 1,64 minute au 57e arrêt. Cette chute peut venir des marges prévues dans les horaires avant le terminus. NB : ces derniers points ne reposent que sur environ 1 000 trains, prudence.

Recommandation : dès le 1er arrêt, les trains ont déjà 1,6 minute de retard en moyenne, soit plus que l'aggravation tout au long du trajet. Il faut donc donner la priorité à la ponctualité au départ et sur les premiers arrêts.

## Question 5 : recommandations pour réduire les retards

1. Faire partir les trains à l'heure. Dès le premier arrêt, les trains ont déjà 1,6 minute de retard en moyenne, et ce retard au départ se retrouve à l'arrivée (1,6 min de retard dès le 1er arrêt, contre environ 0,75 min).

2. Espacer davantage les arrivées aux heures de pointe. Les heures les plus touchées sont 8h (9,3 % de retard), 9h (9,8 %), 17h et 18h (environ 9,5 %), contre environ 6 % en milieu de journée.

3. Concentrer les efforts sur 5 lignes. Les lignes 25, 36N, 36, 96 et 37 représentent à elles seules 29 % des arrivées en retard du réseau. La ligne 37 (Liège–Verviers–Welkenraedt) est la priorité : elle cumule un mauvais taux (16,2 % de retard) et un gros volume.

## Conclusion

J'ai analysé la ponctualité des trains belges à partir des données ouvertes d'Infrabel :  2 067 492  d'arrêts sur septembre 2026, traités avec Python et pandas. L'objectif : comprendre **quand** et **où** les trains prennent du retard.

**Résultats.** 92,6 % des départs et 92,5 % des arrivées sont à l'heure : la ponctualité se joue surtout au départ. Les heures de pointe sont les plus touchées, avec environ 9,5 % de retards à 8-9 h et à 17-18 h, contre environ 6 % en milieu de journée. Le mardi (8,3 %) et le vendredi (8,2 %) sont les pires jours ; le dimanche est le meilleur. Cinq lignes (25, 36N, 36, 96 et 37) concentrent à elles seules 29 % des arrivées en retard. Enfin, le retard s'aggrave peu le long du trajet : de 1,6 à environ 2,4 minutes en moyenne.
NB: Le mardi (8,3 %) et le vendredi (8,2 %) sont les pires jours, mais l’écart entre les jours reste faible comparé à celui entre les heures.

**Recommandation.** Je conseillerais de concentrer les efforts sur la ponctualité au départ aux heures de pointe, et sur ces 5 lignes. La ligne 37 (Liège–Verviers–Welkenraedt) est la priorité : elle cumule un taux de retard élevé (16,2 %) et un fort volume.

**Limites.** Les données ne couvrent qu'un seul mois : les résultats peuvent varier selon la saison (météo, vacances, travaux). Les gares de moins de 1 000 arrêts ont été exclues pour éviter des pourcentages trompeurs. L'explication du mercredi plus calme (temps partiels, horaires scolaires) reste une hypothèse à vérifier. Il faudrait analyser plusieurs mois pour confirmer ces tendances.
